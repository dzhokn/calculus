### Integrals and Area
We know that the definition of an integral is the area "between" the function we're integrating, and the *$x$-axis*. This gives us a method to calculate integrals. 

The simplest way to calculate the integral is to separate the function area into small rectangles and sum the area of these rectangles.

The formula would be:
$$
\int_{x_{min}}^{x_{max}} f(x) dx \approx \sum_{i=1}^{n} x_i \cdot f(x_i)
$$
where:
- $f(x)$ is the function to integrate
- $x_{min}$ and $x_{max}$ are the limits of integration
- $n$ is the number of points (the higher, the more precise approximation)


Let's write a function which does exactly that.


```python
from collections.abc import Callable
import numpy as np
```

```python
def calculate_integral(func: Callable, 
                       x_min: int, 
                       x_max: int, 
                       num_points: int = 100000):
    """
    Calculates a numerical approximation of the definite integral of the provided function
    between the points x_min and x_max.
    The parameter n specifies the number of points at which the function area will be cut into rectangle pieces.

    Args:
        func (Callable): The function to integrate.
        x_min (int): The lower limit of integration.
        x_max (int): The upper limit of integration.
        num_points (int): The number of points at which the integral will be calculated. The higher the number, the more accurate the approximation.

    Returns:
        float: The approximate value of the integral.
    """
    # Generate points between x_min and x_max
    x = np.linspace(x_min, x_max, num_points)

    # Calculate the function values at the points
    y = func(x)

    # Calculate the width of the micro-rectangles (the distance between the last and first point divided by the number of rectangles)
    delta_x = (x_max - x_min) / num_points

    # Calculate and sum the area of the microscopically small rectangles (dx * y)
    return np.sum(delta_x * y)
```

```python
print(calculate_integral(lambda x: x ** 2, 0, 1))           # Should be close to 0.333
print(calculate_integral(lambda x: np.sin(x ** 2), 0, 5))   # Should be close to 0.528
```

Let's apply our insight to finding the area of a circle. We know the equation of a circle is not a function (it's more like two functions). We can, however be clever. One way is to integrate both of the functions and add them together. Another way is to integrate one and double the area. 

**Note:** We're trying to find the total area of the circle, there is **no negative area** in this particular case.

Another, even more clever way is to look at a quarter of the circle. This is convenient because we may look at the quadrant where $x > 0$ and $y > 0$. So, we'll need to find the area between:
1. $x \ge 0$
2. $y \ge 0$
3. The circle $x^2 + y^2 \le R^2$ (let's fix the radius to be 1)

$\Rightarrow y = \sqrt{R^2 - x^2} = \sqrt{1 - x^2}$

After all of this, we'll need to multiply the result by 4.

$$ S = 4 \int_0^1\sqrt{1 - x^2}dx $$

```python
circle_piece_area = calculate_integral(lambda x: np.sqrt(1 - x ** 2), 0, 1)
total_area = 4 * circle_piece_area
print(total_area)
```

### Fundamental theorem of calculus

The [fundamental theorem of calculus](https://en.wikipedia.org/wiki/Fundamental_theorem_of_calculus) is a theorem that links the concept of differentiating a function (calculating its slopes, or rate of change at every point on its domain) with the concept of integrating a function (calculating the area under its graph, or the cumulative effect of small contributions). 

Roughly speaking, the two operations (differentiating and integrating) can be thought of as **inverses of each other**.

### Fundamental theorem of calculus (part 2)
Conversely, the second part of the theorem, the second fundamental theorem of calculus, states that the integral of a function $f$ over a fixed interval is equal to the change of any antiderivative $F$ between the ends of the interval. This greatly simplifies the calculation of a definite integral provided an antiderivative can be found.

$$ \int_a^b f(x) dx = F(b) - F(a) $$

where $F$ is the antiderivitive of $f(x)$.
